# Phase 2 — EDA (figures saved to reports/figures/)

```python
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
df = pd.read_csv('data/processed/charging_cleaned.csv', parse_dates=['StartDatetime'])
print(df['Demand'].describe().to_string())
print(df.groupby('ChargerType')['Demand'].mean())
print(df.groupby('Location')['Demand'].mean().sort_values(ascending=False))
print(df[['Duration','Demand']].corr())
```

```python
plt.hist(df['Demand'], bins=50); plt.title('Demand (right-skewed)'); plt.show()
sns.boxplot(data=df, x='ChargerType', y='Demand'); plt.show()
df.groupby('Location')['Demand'].mean().sort_values().plot(kind='barh', figsize=(6,5)); plt.show()
df.set_index('StartDatetime').resample('D')['Demand'].mean().plot(figsize=(10,3)); plt.show()
sns.heatmap(df[['UserID','ChargerID','ChargerCompany','ChargerType','Duration','Demand']].corr(), annot=True); plt.show()
```
