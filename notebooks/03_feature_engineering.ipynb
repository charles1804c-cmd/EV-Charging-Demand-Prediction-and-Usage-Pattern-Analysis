# Phase 3 — Feature Engineering (calls src/feature_engineering.py)

```python
from src.feature_engineering import build_features, encode_and_split
import pandas as pd
df = pd.read_csv('data/processed/charging_cleaned.csv', parse_dates=['StartDatetime','EndDatetime'])
feat = build_features(df)
print('rare grouped:', feat.attrs.get('rare_locations'))
print(feat[['hour','day_of_week','is_weekend','time_of_day','user_past_avg_demand','charger_past_avg_demand']].head())
```

```python
train, test, arts = encode_and_split(feat)
train.to_csv('data/processed/modeling_train.csv', index=False)
test.to_csv('data/processed/modeling_test.csv', index=False)
import pickle; pickle.dump(arts, open('data/processed/feature_artifacts.pkl','wb'))
print('split:', arts['split_datetime'], train.shape, test.shape)
```
